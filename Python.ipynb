```python
from pathlib import Path
import os

def readfileandfolder():
    path = Path('')
    items = list(path.rglob('*'))
    for i, itmes in enumerate(items):
        print(f"{i+1} : {items}")


def createfile():
    try:

     readfileandfolder()
     name = input("Please tell your file name :-")

     p = Path(name)
     if not p.exists():
        with open(p,"w") as fs:
           data = input("What you want to write in yhis file :-")
           fs.write(data)

        print(F"FILE CREATED SUCCESSFULLY.")
     else:
        print("This file already exist")

    except Exception as err:
       print(f"an error occured as {err}")


def readfile():
   try:

     readfileandfolder()
     name = input("Which file you want to read :-")
     p = Path(name)
     if p.exists() and p.is_file():
        with open(p,"r") as fs:
           data = fs.read()
           print(data)
        print("readed successfully")
     else:
        print("This file does not exist")
   except Exception as err:
      print(f"an error occures as {err}")


def updatefile():
   try:
      readfileandfolder()
      name = input("Which file you want to update :-")
      p = Path(name)
      if p.exists() and p.is_file():
         print("Press 1 for changing the name of your file :-")
         print("Press 2 for overwriting the data of your file :-")
         print("Press 3 for appending the content in your file :-")

         res = int(input("Tell your response :-"))

         if res == 1:
            name2 = input("Tell your new file name :-")
            p2 = Path(name2)
            p.rename(p2)

         if res == 2:
            with open(p,'w') as fs:
               data = input("Tell what you want to write this is overwrite the data :-")
               fs.write(data)

         if res == 3:
            with open(p,'a') as fs:
               data = input("Tell what you want to append :-")
               fs.write(" "+data)

   except Exception as err:
      print(f"an error occures as {err}")


def deletefile():
   try:
      readfileandfolder()
      name = input("Which file you want to delete :-")
      p = Path(name)
      if p.exists() and p.is_file():
        os.remove(name)

        print("File remove successfully")
      else:
         print("No such file exist")
   except Exception as err:
      print(f"an error occured as {err}")



print("Press 1 for creating a file")
print("Press 2 for reading a file")
print("Press 3 for updaring a file")
print("Press 4 for deletion a file")

check = int(input("Please tell your response :-"))

if check == 1:
    createfile()

if check == 2:
   readfile()

if check == 3:
   updatefile()

if check == 4:
   deletefile()
```

**Abstraction**

```python
from abc import ABC, abstractmethod
class abstract(ABC):
  @abstractmethod
  def perimeter(self):
    pass
  @abstractmethod
  def area(self):
    pass


class Square(abstract):
  def __init__(self,sides):
    self.sides = sides

class Circle(abstract):
  def __init__(self,radius):
    self.radius = radius
  def perimeter(self):
     print("I have created.")
  def area(self):
     print("I have created this.")

obj = Circle(12)
obj.area()
```

```python
class Animal:
  def __init__(self,name,age):
    self.name = name
    self.age = age
  def __str__(self):
    return f"Hello how are you and your name is {self.name} "

  def __add__(self,other):
    sum = 0
    for i in other:
      sum = sum + i.age
    return f"Your sum of ages are {self.age + sum}"

obj = Animal("Lion",12)
obj2 = Animal("Tiger",13)
obj3 = Animal("Fish",14)
print(obj+(obj2,obj3))
```

**Decorator**

```python
class Animal:
  @property
  def show(self):
    print("How are you?")

obj = Animal()
obj.show
```

```python
def decorate(func):
  def wrapper():
    print("I will print myself before the function hello.")
    func()
    print("I will print myself after the function.")
  return wrapper

@decorate
def hello():
  print("Hello I am a girl.")

hello()
```

```python
def decorate(func):
  def wrapper(a,b):
    print("The addition to your numbers are:-")
    func(a,b)
    print("Thank you I hope you liked it.")
  return wrapper

@decorate
def addition(a,b):
  print(f"Your total is {a + b}")

addition(12,34)
```

```python
def decorate(func):
  def wrapper(*args,**kwargs):
    print("The addition to your numbers are:-")
    func(*args,**kwargs)
    print("Thank you I hope you liked it.")
  return wrapper

@decorate
def addition(a,b,c,d):
  print(f"Your total is {a + b +c +d}")

addition(12,34,23,14)
```

**Args and Kwargs**

```python
def addition(*args):
  sum = 0
  for i in args:
    sum = sum + i
  print(sum)

addition(6,12,24,48)
```

```python
def addition(**kwargs):
  print(kwargs)

addition(a=12,b=56,c=72)
```

```python
def information(**kwargs):
  print("Your information is :-")
  for i in kwargs:
    print(f"{i} : {kwargs[i]}")

information(name = "Amna",age = 21,Designation = "AI/ML")
```

**Lambda Functions**

```python
addition = lambda a,b : a+b
print(addition(12,12))
```

```python
evenodd = lambda a : "even" if a%2==0 else "odd"
print(evenodd(13))
```

**MAP**

```python
a = [1,2,3,4,5]
result = map(lambda x : x*2,a)
print(list(result))
```

```python
a = [1,2,3,4,5]
def double(x):
  return x*2

result = map(double,a)
print(list(result))
```

**Filter**

```python
def even(x):
  if x%2==0:
    return True
  else:
    return False

a = [1,2,3,4,5,6,7,8,9]
result = filter(even,a)
print(list(result))
```

```python
a = [1,2,3,4,5,6,7,8,9]
result = filter(lambda x : True if x%2==0 else False ,a)
print(list(result))
```
