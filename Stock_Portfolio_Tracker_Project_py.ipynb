```python
import csv
from datetime import datetime

# Hardcoded dictionary of stock prices
stock_prices = {
    "AAPL": 180,
    "TSLA": 250,
    "MSFT": 320,
    "GOOGL": 140,
    "AMZN": 170
}

def portfolio_tracker():
    portfolio = {}
    print("📈 Welcome to Stock Portfolio Tracker 📈")
    print("Available stocks:", ", ".join(stock_prices.keys()))
    print("Enter 'done' when finished.\n")

    # User input loop
    while True:
        stock = input("Enter stock symbol (AAPL/TSLA/MSFT/GOOGL/AMZN): ").upper()
        if stock == "DONE":
            break
        if stock not in stock_prices:
            print("❌ Invalid stock symbol. Try again.")
            continue

        try:
            qty = int(input(f"Enter quantity of {stock}: "))
        except ValueError:
            print("❌ Quantity must be a number.")
            continue

        portfolio[stock] = portfolio.get(stock, 0) + qty

    # Calculate total investment
    total_value = 0
    print("\n📊 Portfolio Summary 📊")
    for stock, qty in portfolio.items():
        value = stock_prices[stock] * qty
        total_value += value
        print(f"{stock} → {qty} shares × ${stock_prices[stock]} = ${value}")

    print(f"\n💰 Total Investment Value = ${total_value}")

    # Save to file option
    save = input("\nDo you want to save portfolio to file? (y/n): ").lower()
    if save == "y":
        filename = f"portfolio_{datetime.now().strftime('%Y%m%d_%H%M%S')}.csv"
        with open(filename, "w", newline="") as f:
            writer = csv.writer(f)
            writer.writerow(["Stock", "Quantity", "Price", "Value"])
            for stock, qty in portfolio.items():
                writer.writerow([stock, qty, stock_prices[stock], stock_prices[stock]*qty])
            writer.writerow(["TOTAL", "", "", total_value])
        print(f"✅ Portfolio saved successfully as {filename}")

# Run the tracker
portfolio_tracker()
```
