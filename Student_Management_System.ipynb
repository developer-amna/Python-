```python
class Student:
    def __init__(self, roll_no, name, grade):
        self.roll_no = roll_no
        self.name = name
        self.grade = grade

class StudentManagement:
    def __init__(self):
        self.students = []

    def add_student(self, roll_no, name, grade):
        new_student = Student(roll_no, name, grade)
        self.students.append(new_student)
        print("Student added successfully!")

    def display_students(self):
        print("\n--- Student Records ---")
        for student in self.students:
            print(f"Roll No: {student.roll_no} | Name: {student.name} | Grade: {student.grade}")

    def update_student(self, roll_no, new_name, new_grade):
        for student in self.students:
            if student.roll_no == roll_no:
                student.name = new_name
                student.grade = new_grade
                print("Student record updated successfully!")
                return
        print("Student not found!")

    def delete_student(self, roll_no):
        for student in self.students:
            if student.roll_no == roll_no:
                self.students.remove(student)
                print("Student record deleted successfully!")
                return
        print("Student not found!")

def main():
    sm = StudentManagement()

    while True:
        print("\n--- Student Management Menu ---")
        print("1. Add Student")
        print("2. Display Students")
        print("3. Update Student")
        print("4. Delete Student")
        print("5. Exit")

        choice = input("Enter your choice: ")

        if choice == "1":
            roll_no = int(input("Enter Roll No: "))
            name = input("Enter Name: ")
            grade = input("Enter Grade: ")
            sm.add_student(roll_no, name, grade)

        elif choice == "2":
            sm.display_students()

        elif choice == "3":
            roll_no = int(input("Enter Roll No to update: "))
            new_name = input("Enter New Name: ")
            new_grade = input("Enter New Grade: ")
            sm.update_student(roll_no, new_name, new_grade)

        elif choice == "4":
            roll_no = int(input("Enter Roll No to delete: "))
            sm.delete_student(roll_no)

        elif choice == "5":
            print("Exiting program...")
            break

        else:
            print("Invalid choice!")

if __name__ == "__main__":
    main()
```
