***STAR PATTERNS***

**Right-Angled Triangle(Left-Aligned)**

```python
n = 10
for i in range (1, n+1):
  print ("* " * i)
```

**Right-Angled Triangle(Right-Aligned)**

```python
n = 10
for i in range (1, n+1):
  print("  " * (n-i) + "* " * i)
```

**Pyramid Pattern (Centered)**

```python
n = 10
for i in range (1, n+1):
  print(" " * (n-i) + "* " * i)
```

**Inverted Pyramid**

```python
n = 10
for i in range (n , 0 , -1):
  print(" " * (n-i) + "* " * i)
```

**Diamond Pattern**

```python
n = 10
for i in range (1, n+1):
  print(" " * (n-i) + "* " * i)
for i in range (n-1 , 0 , -1):
  print(" " * (n-i) + "* " * i)
```

**Square Pattern**

```python
n = 10
for i in range (1, n+1):
  print ("* " *n )
```
