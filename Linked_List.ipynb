**Creating Singly linked list**

```python
class Node:
    def __init__(self, rollno, name):
        self.rollno = rollno
        self.name = name
        self.link = None


class StudentList:
    def __init__(self):
        self.start = None

    def insert(self, rollno, name):
        temp = Node(rollno, name)
        if self.start is None:
            self.start = temp
        else:
            current = self.start
            while current.link is not None:
                current = current.link
            current.link = temp

    def display(self):
        if self.start is None:
            print("List is empty")
            return
        else:
            rec = 1
            current = self.start
            while current is not None:
                print(f"Record#{rec}")
                print(f"Roll Number : {current.rollno}")
                print(f"Name : {current.name}")
                current = current.link
                rec += 1


def main():
    obj = StudentList()
    rec = 1
    while rec <= 5:
        print(f"Enter Record #{rec}")
        m = int(input("Enter Roll No : "))
        nm = input("Enter Name : ")
        obj.insert(m, nm)
        rec += 1

    obj.display()


if __name__ == "__main__":
    main()
```

**Operations on Singly Linked List**

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.link = None


class LinkedList:
    def __init__(self):
        self.start = None

    # Insert a node at the beginning of the list
    def insert_begin(self):
        data = float(input("Enter the data value for a node: "))
        temp = Node(data)
        if self.start is None:
            self.start = temp
        else:
            temp.link = self.start
            self.start = temp

    # Insert a node at the end of the list
    def insert_end(self):
        data = float(input("Enter the data value for a node: "))
        temp = Node(data)
        if self.start is None:
            self.start = temp
        else:
            current = self.start
            while current.link is not None:
                current = current.link
            current.link = temp

    # Insert a node after a specific value
    def insert_location(self):
        if self.start is None:
            print("List is empty.")
            return
        n = float(input("Enter the value to search: "))
        current = self.start
        found = False
        while current is not None:
            if current.data == n:
                found = True
                break
            current = current.link
        if found:
            data = float(input("Enter the data value for a node: "))
            temp = Node(data)
            temp.link = current.link
            current.link = temp
        else:
            print("Value not found")

    # Delete the first node
    def delete_begin(self):
        if self.start is None:
            print("List is empty")
            return
        current = self.start
        self.start = current.link
        print("Data of first node is:", current.data)
        print("First node is deleted")

    # Delete the last node
    def delete_end(self):
        if self.start is None:
            print("List is empty")
            return
        current = self.start
        previous = None
        while current.link is not None:
            previous = current
            current = current.link
        print("Data of last node is:", current.data)
        if previous:
            previous.link = None
        else:
            self.start = None
        print("Last node is deleted")

    # Delete a node at a specific location (by value)
    def delete_location(self):
        if self.start is None:
            print("List is empty")
            return
        n = float(input("Enter a value to search: "))
        current = self.start
        previous = None
        found = False
        while current is not None:
            if current.data == n:
                found = True
                break
            previous = current
            current = current.link
        if found:
            if previous:
                previous.link = current.link
            else:
                self.start = current.link
            print("Node is deleted")
        else:
            print("Value not found")

    # Bubble sort the linked list
    def bubble_sort(self):
        if self.start is None:
            print("List is empty")
            return
        size = self.count()
        for u in range(size, 1, -1):
            current = self.start
            for i in range(1, u):
                nxt = current.link
                if nxt and current.data > nxt.data:
                    current.data, nxt.data = nxt.data, current.data
                current = nxt

    # Count the number of nodes
    def count(self):
        current = self.start
        count = 0
        while current is not None:
            count += 1
            current = current.link
        return count

    # Search for a value in the list
    def search(self):
        if self.start is None:
            print("List is empty")
            return
        value = float(input("Enter a value to search: "))
        current = self.start
        found = False
        while current is not None:
            if current.data == value:
                found = True
                break
            current = current.link
        if found:
            print("Value found in the list")
        else:
            print("Value not found in the list")

    # Display all nodes
    def display(self):
        if self.start is None:
            print("List is empty")
            return
        current = self.start
        while current is not None:
            print(current.data)
            current = current.link


def main():
    obj = LinkedList()
    loop = True
    while loop:
        print("\n01- Insert a node at the beginning")
        print("02- Insert a node at the end")
        print("03- Insert a node at a specific location")
        print("04- Delete a node at the beginning")
        print("05- Delete a node at the end")
        print("06- Delete a node at a specific location")
        print("07- Sort the values of the list")
        print("08- Search the value from a list")
        print("09- Display data of nodes")
        print("10- Count nodes of the list")
        print("11- Exit")
        op = int(input("Enter your option [1-11]: "))

        if op == 1:
            obj.insert_begin()
        elif op == 2:
            obj.insert_end()
        elif op == 3:
            obj.insert_location()
        elif op == 4:
            obj.delete_begin()
        elif op == 5:
            obj.delete_end()
        elif op == 6:
            obj.delete_location()
        elif op == 7:
            obj.bubble_sort()
        elif op == 8:
            obj.search()
        elif op == 9:
            obj.display()
        elif op == 10:
            print("Total nodes are:", obj.count())
        elif op == 11:
            loop = False
        else:
            print("Invalid option")


if __name__ == "__main__":
    main()
```
