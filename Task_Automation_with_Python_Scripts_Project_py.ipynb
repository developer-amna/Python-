```python
import requests
import re

def scrape_title(url, output_file):
    response = requests.get(url)
    html = response.text

    # Regex to find <title> tag
    match = re.search(r"<title>(.*?)</title>", html, re.IGNORECASE)
    if match:
        title = match.group(1)
        with open(output_file, "w") as f:
            f.write(title)
        print(f"Page title saved to {output_file}: {title}")
    else:
        print("No title found.")

# Example usage
scrape_title("https://www.python.org", "title.txt")
```
