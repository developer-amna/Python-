```python
import random

def chatbot():
    print("🤖 Welcome to Smart Chatbot! (type 'bye' to exit)")

    # Predefined replies dictionary
    replies = {
        "hello": ["Hi!", "Hello there!", "Hey, nice to see you!"],
        "how are you": ["I'm fine, thanks!", "Doing great, how about you?", "All good here!"],
        "bye": ["Goodbye!", "See you later!", "Take care!"],
        "what is your name": ["I'm your chatbot!", "You can call me PyBot.", "My name is SmartBot!"]
    }

    while True:
        user_input = input("You: ").lower()

        if user_input in replies:
            # Random reply from list
            print("Bot:", random.choice(replies[user_input]))
            if user_input == "bye":
                break
        else:
            print("Bot: Sorry, I don't understand that. Try 'hello', 'how are you', or 'bye'.")

# Run the chatbot
chatbot()
```
